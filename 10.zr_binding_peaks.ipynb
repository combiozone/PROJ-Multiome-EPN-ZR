```r
library(ggplot2)
library(ggrepel)
library(ggpubr)
library(data.table)
```

```r
df <- fread('database/zrfus_binding_peaksig.closetgene_zrfus.tsv', data.table=F)
zr_gene <- readLines('database/ZRFUS_signature_genes.93.mouse.gene')
```

```r
ggplot(df, aes(x=peaksig_x, y=peaksig_y) ) +
            geom_point(size=3, color = "#cdcdcd", alpha=.7) + 
            xlim(0, 1) + ylim(0, 1) +
            geom_abline(linetype = "dashed") +
            theme_linedraw() + 
            theme(panel.grid.major = element_blank(), panel.grid.minor = element_blank()) +
            labs(title='Accessibility of ZFTA-RELA\n binding sites', y='RGCs (Mean peak signal)', x='Neurons (Mean peak signal)') +
            theme(plot.title = element_text(face='bold', hjust = 0.5)) +
            geom_point(data=subset(df, closet_genes %in% zr_gene), size=3, color = "#C92100") +
            geom_text_repel(data = subset(df, closet_genes %in% c('Aif1l','Jag1','Gli2','Rela')), aes(label = closet_genes), size = 8)
```
